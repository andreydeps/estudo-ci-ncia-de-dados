```python
import google.generativeai as genai
from google.colab import userdata

# 1. Busca a chave do Secrets
minha_chave = userdata.get('GEMINI_API_KEY')

# 2. Configura a biblioteca
genai.configure(api_key=minha_chave)

# 3. Inicializa o modelo que você viu que está disponível
# Se o sistema listou 'gemini-2.5-flash', usamos exatamente assim
model = genai.GenerativeModel('gemini-2.5-flash')

# 4. O Teste do Oi
response = model.generate_content("Dê um oi curto para um estudante de economia da UFSC.")

print(response.text)
```

```python
# Nossa lista de dicionários com dados de exportação de SC
# Pense nisso como uma tabela: cada dicionário {} é uma linha
dados_sc = [
    {"produto": "Motores Elétricos", "valor_usd": 850000000, "setor": "Metal-Mecânico"},
    {"produto": "Carne Suína", "valor_usd": 1400000000, "setor": "Agroindústria"},
    {"produto": "Frango", "valor_usd": 1600000000, "setor": "Agroindústria"},
    {"produto": "Compressores", "valor_usd": 600000000, "setor": "Metal-Mecânico"},
    {"produto": "Móveis", "valor_usd": 350000000, "setor": "Madeireiro"}
]

print(f"Sucesso! Temos {len(dados_sc)} categorias carregadas na memória.")
```

```python
import pandas as pd

# Transforma a lista que você criou em uma tabela (DataFrame)
df = pd.DataFrame(dados_sc)

# Exibe a tabela bonitona
df
```

```python
# 1. Transformamos nossa lista em um texto para a IA ler
texto_dados = str(dados_sc)

# 2. Criamos o comando (Prompt) usando os seus conhecimentos de Economia
prompt = f"""
Você é um economista sênior especialista na economia de Santa Catarina.
Analise os dados de exportação abaixo:
{texto_dados}

Com base nesses dados, faça o seguinte:
1. Identifique o setor com maior volume financeiro.
2. Explique um risco econômico para esse setor (ex: câmbio, demanda externa).
3. Sugira uma estratégia de diversificação para SC baseada em complexidade econômica.
Responda de forma técnica, mas direta.
"""

# 3. Enviamos para o modelo 2.5-flash que funcionou antes
response = model.generate_content(prompt)

# 4. Exibimos o resultado
print("--- ANÁLISE DO AGENTE ECONOMISTA ---")
print(response.text)
```
